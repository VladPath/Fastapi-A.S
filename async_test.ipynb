```python
import aiohttp
import asyncio
```

```python
async def get_data(i:int, endpoint:str):
    print(f"Начал выполнение функции {i}")
    url = f"http://localhost:8002/{endpoint}/{i}"
    async with aiohttp.ClientSession() as session:
        async with session.get(url) as resp:
            print(f"Завершил выполнение функции {i}")
```

```python
await asyncio.gather(
     *[get_data(i,'async') for i in range(600)]
)
```
